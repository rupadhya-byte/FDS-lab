```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Automobile_data.csv")

# Replace '?' with NaN and convert 'price' column to numeric
df['price'] = pd.to_numeric(df['price'], errors='coerce')

# Display first 5 rows
print(df.head())

# Diagnose the 'price' column dtype and unique values after conversion
print(f"'price' column dtype: {df['price'].dtype}")
if df['price'].dtype == 'object':
    print(f"Unique non-numeric values in 'price': {df['price'][df['price'].apply(lambda x: not isinstance(x, (int, float)) or pd.isna(x))].unique()}")
```

```python
plt.figure(figsize=(8,5))
df["price"].hist(bins=20)
plt.title("Distribution of Automobile Prices")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset and preprocess 'price' column
df = pd.read_csv("Automobile_data.csv")
df['price'] = pd.to_numeric(df['price'], errors='coerce')

avg_price = df.groupby("make")["price"].mean()

plt.figure(figsize=(10,5))
avg_price.plot(kind="bar")
plt.title("Average Price by Company")
plt.xlabel("Company")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["horsepower"], df["price"])
plt.title("Horsepower vs Price")
plt.xlabel("Horsepower")
plt.ylabel("Price")
plt.show()
```

```python
plt.figure(figsize=(6,5))
plt.boxplot(df["price"].dropna())
plt.title("Box Plot of Prices")
plt.ylabel("Price")
plt.show()
```

```python
fuel_count = df["fuel-type"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(fuel_count,
        labels=fuel_count.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Fuel Type Distribution")
plt.show()
```

```python
sorted_df = df.sort_values("price")

plt.figure(figsize=(10,5))
plt.plot(sorted_df["price"])
plt.title("Price Trend")
plt.xlabel("Automobile Index")
plt.ylabel("Price")
plt.show()
```

```python
fuel = df["fuel-type"].value_counts()

plt.figure(figsize=(6,5))
fuel.plot(kind="bar")
plt.title("Number of Cars by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Count")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["horsepower"].hist(bins=15)
plt.title("Horsepower Distribution")
plt.xlabel("Horsepower")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["engine-size"], df["price"])
plt.title("Engine Size vs Price")
plt.xlabel("Engine Size")
plt.ylabel("Price")
plt.show()
```

```python
df['average-mileage'] = (df['city-mpg'] + df['highway-mpg']) / 2
avg_mileage = df.groupby("make")["average-mileage"].mean()

plt.figure(figsize=(10,5))
avg_mileage.plot(kind="barh")
plt.title("Average Mileage by make")
plt.xlabel("Average Mileage")
plt.ylabel("make")
plt.show()
```

```python
plt.figure(figsize=(8,5))
df["engine-size"].hist(bins=15)
plt.title("Engine Size Distribution")
plt.xlabel("Engine Size")
plt.ylabel("Frequency")
plt.show()
```
