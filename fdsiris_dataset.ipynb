```python
import pandas as pd
```

```python
df = pd.read_csv("Iris.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
numeric = df.select_dtypes(include='number')
numeric
```

```python
mean = numeric.mean()
print("Mean")
print(mean)
```

```python
median = numeric.median()
print("Median")
print(median)
```

```python
mode = numeric.mode()

print("Mode")
print(mode)
```

```python
summary = pd.DataFrame({
    "Mean": numeric.mean(),
    "Median": numeric.median(),
    "Mode": numeric.mode().iloc[0]
})

print(summary)
```
